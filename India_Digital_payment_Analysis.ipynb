```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

file_path = r"C:\Users\Sendhamil selvan\OneDrive\New folder\Documents\UPI-Fraud_Analysis\data\raw\India_Digital_Payments_Data.xlsx"

df = pd.read_excel(file_path, sheet_name="UPI_Monthly")

# Clean column names
df.columns = df.columns.astype(str).str.strip()

print(df.columns.tolist())
print("Shape:", df.shape)
display(df.head())
```

```python
# Check missing values before cleaning
print("Missing values before cleaning:")
print(df.isnull().sum())

# Remove duplicate rows
df = df.drop_duplicates().copy()

# Convert Month to datetime
df["Month"] = pd.to_datetime(df["Month"], errors="coerce")

# Remove rows where Month could not be interpreted
df = df.dropna(subset=["Month"])

# Sort by month
df = df.sort_values("Month").reset_index(drop=True)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

display(df.head())
```

```python
# Select important columns
analysis_df = df[
    ["Month", "Volume (Mn)", "Value (Rs Cr)", "Avg Ticket Size (Rs)"]
].copy()

# Filter months with transaction volume greater than 10,000 million
high_volume = analysis_df[analysis_df["Volume (Mn)"] > 10000]

# Calculate month-on-month volume growth
df["Calculated MoM Volume %"] = df["Volume (Mn)"].pct_change() * 100

display(high_volume.head())
display(df[["Month", "Volume (Mn)", "Calculated MoM Volume %"]].tail())
```

```python
plt.figure(figsize=(12, 5))
plt.plot(df["Month"], df["Volume (Mn)"])
plt.title("Monthly UPI Transaction Volume")
plt.xlabel("Month")
plt.ylabel("Volume (Millions)")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
